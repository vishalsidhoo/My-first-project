```python
# Create grouped bar chart
plt.figure(figsize=(12,6))
# Convert data to DataFrame using reset_index()
data = df.groupby(["Disease Name", "Disease Category"])["Average Treatment Cost (USD)"].mean().sort_values(ascending=False).reset_index()
sns.barplot(x="Disease Name", y="Average Treatment Cost (USD)", hue="Disease Category", data=data)

plt.xticks(rotation=45)
plt.title("Average Treatment Cost by Disease & Country")
plt.ylabel("Avg Treatment Cost (USD)")
plt.xlabel("Disease Name")
plt.legend(title="Country")
plt.show()
```

# Country Vs Improvement Rate By(%)

```python
data=df.groupby("Country")["Improvement in 5 Years (%)"].mean().sort_values(ascending=False)
```

```python
# Reset the index of the 'data' Series to convert it into a DataFrame
data = df.groupby("Country")["Improvement in 5 Years (%)"].mean().sort_values(ascending=False).reset_index()

# Now you can use the DataFrame in the barplot function
plt.figure(figsize=(12,6))
sns.barplot(x="Country", y="Improvement in 5 Years (%)", data=data)

plt.xticks(rotation=45)
plt.title("Average Improvement in 5 Years by Country") # Changed title to reflect the data
plt.ylabel("Avg Improvement in 5 Years (%)") # Changed y-axis label to reflect the data
plt.xlabel("Country") # Changed x-axis label to reflect the data
plt.legend(title="Country") # This might not be needed as country is already on x-axis
plt.show()
```

Disease Prevalence Across Countries


```python
plt.figure(figsize=(12, 6))
top_countries = df.groupby("Country")["Prevalence Rate (%)"].mean().sort_values(ascending=False).head(10)
sns.barplot(x=top_countries.index, y=top_countries.values, palette="Reds")
plt.xticks(rotation=45)
plt.title("Top 10 Countries with Highest Disease Prevalence")
plt.show()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(14, 6))

sns.scatterplot(x=df["Doctors per 1000"], y=df["Mortality Rate (%)"], ax=ax[0], color="blue", alpha=0.6)
ax[0].set_title("Doctors per 1000 vs Mortality Rate")

sns.scatterplot(x=df["Hospital Beds per 1000"], y=df["Mortality Rate (%)"], ax=ax[1], color="green", alpha=0.6)
ax[1].set_title("Hospital Beds per 1000 vs Mortality Rate")

plt.show()
```
